The initial model included 14 predictors and interaction terms. Although it achieved an R² of 0.803, VIF analysis revealed substantial multicollinearity, particularly among points, shooting attempts, and interaction variables.

```python
import pandas as pd
import statsmodels.api as sm
# data loading
df = pd.read_csv('//content/nba_team_stats_00_to_23.csv')

#clean columns
df.columns = df.columns.str.strip().str.lower()


# 3. Filter seasons (2013–2023)
# season looks like "2012-13", so extract starting year
df['season_start'] = df['season'].str[:4].astype(int)
df = df[df['season_start'] >= 2013].copy()

# 4. Target variable (win %)

df['win_pct'] = df['wins'] / df['games_played']


# 5. Interaction terms
df['three_interaction'] = df['three_point_percentage'] * df['three_pointers_attempted']
df['fg_interaction'] = df['field_goal_percentage'] * df['field_goals_attempted']


# 6. Features (clean + non-redundant)

features = [
    'points',
    'assists',
    'rebounds',
    'steals',
    'blocks',
    'turnovers',
    'three_point_percentage',
    'three_pointers_attempted',
    'field_goal_percentage',
    'field_goals_attempted',
    'free_throw_percentage',
    'free_throw_attempted',
    'three_interaction',
    'fg_interaction'
]

#prepare data
df_model = df[features + ['win_pct']].dropna()

X = df_model[features]
y = df_model['win_pct']

# Add intercept
X = sm.add_constant(X)


# 8. Fit model

model = sm.OLS(y, X).fit()


# 9. Results

print(model.summary())
```

```python

# 10. Correlation matrix

corr_matrix = df_model[features].corr()

# Print full matrix
print(corr_matrix)

# Optional: focus on strong correlations
high_corr = corr_matrix[(corr_matrix.abs() > 0.8) & (corr_matrix.abs() < 1.0)]
print("\nHigh correlations (> 0.8):\n", high_corr)
```

```python

# 11. VIF calculation

from statsmodels.stats.outliers_influence import variance_inflation_factor

# Need a DataFrame without the target
X_vif = sm.add_constant(df_model[features])

vif_data = pd.DataFrame()
vif_data["feature"] = X_vif.columns
vif_data["VIF"] = [variance_inflation_factor(X_vif.values, i)
                   for i in range(X_vif.shape[1])]

print(vif_data)
```
